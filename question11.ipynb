```python
def compute_mean(values):
    return sum(values) / len(values)

compute_mean([2, 4, 6, 8])
```

This function calculates the arithmetic mean (average) of a list of numbers by summing all the values and dividing by the count of values.

The mean can be expressed using summation notation:

$$\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i$$

Or equivalently, written out explicitly:

$$\bar{x} = \frac{x_1 + x_2 + \cdots + x_n}{n}$$
